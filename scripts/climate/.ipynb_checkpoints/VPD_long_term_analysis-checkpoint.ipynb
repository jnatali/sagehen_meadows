```python
from pathlib import Path
from datetime import datetime
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go
```

```python
### Functions for calculating VPD transcribed from the R package 'plantecophys'
# Will move to separate python file in future

def esat(TdegC, Pa=101):
    """
    TdegC : temperature (C)
    Pa : atmospheric pressure (kPa)
    """
    a = 611.21
    b = 17.502
    c = 240.97
    f = 1.0007 + 3.46 * (10 ** -8) * Pa * 1000
    esatval = f * a * (np.exp(b * TdegC / (c + TdegC)))    
    return esatval
    
def RH_to_VPD(RH, TdegC, Pa=101):
    """
    RH : relative humidity (%)
    TdegC : temperature (C)
    Pa : atmospheric pressure (kPa)
    ---
    Returns VPD in kPa
    """
    esatval = esat(TdegC, Pa)
    e = (RH / 100) * esatval
    VPD = (esatval - e) / 1000
    return VPD
```

```python
### Utility functions

def calculate_VPD_series(df:pd.DataFrame, T_name:str, RH_name:str):
    """
    df: pd.DataFrame containing columns named T_name and RH_name
    ---
    Returns Series of VPD values
    """
    return df.apply(lambda row: RH_to_VPD(row[RH_name], row[T_name]), axis=1)

def group_vpd_by_date(df:pd.DataFrame):
    """
    df: pd.DataFrame containing columns named "datetime", "vpd"
    ---
    Returns the resulting DataFrame after aggregating by date with columns "min_vpd", "max_vpd", "avg_vpd", "DOY", "year"
    """
    
    return_df = df.groupby(df['datetime'].apply(lambda val: val.date()), as_index=False).agg(min_vpd=('vpd', 'min'), max_vpd=('vpd', 'max'), avg_vpd=('vpd', 'mean'))
    return_df['DOY'] = return_df['datetime'].apply(lambda val: int(val.strftime('%j')))
    return_df['year'] = return_df['datetime'].apply(lambda val: int(val.strftime('%Y')))
    
    return return_df

def calculate_median_df(df:pd.DataFrame, vpd_column:str, year_list):
    """
    df: pd.DataFrame containing columns named vpd_column, "year", "DOY"
    vpd_column: name of column to aggregate
    year_list: list of years to filter df by
    ---
    Returns DataFrame with columns "DOY", "doy_median"
    """
    filtered_df = df[df["year"].apply(lambda val: val in year_list)]
    filtered_df = filtered_df.groupby('DOY', as_index=False).agg(doy_median=(vpd_column, 'median'))
    return filtered_df

def create_annual_lineplot(df:pd.DataFrame, vpd_column:str, year_list, highlight_year=None, highlight_color='#c73620', highlight_dash=None):
    """
    Creates a plotly figure of a VPD time series, with DOY as the x-axis and VPD as the y-axis.
    ---
    df must have 'year', 'DOY', and vpd_column as columns
    highlight_year: str
    """
    filtered_df = df[df['year'].apply(lambda val: val in year_list)]
    
    fig = px.line(filtered_df, x='DOY', y=vpd_column, color='year')
    if highlight_year:
        traces = list(fig.data)
        fig.data = [t for t in fig.data if t.name != highlight_year] + [t for t in fig.data if t.name == highlight_year]
        for trace in traces:
            if trace.name != highlight_year:
                trace.line.color = 'lightgrey'
            else:
                if highlight_dash:
                    trace.line.dash = highlight_dash
                trace.line.color = highlight_color
    # These lines roughly denote the growing season
    fig.add_vline(x='121', line_dash='dash')
    fig.add_vline(x='305', line_dash='dash')
    
    return fig
```

```python
### Transformations for full dataset
weather_full = weather_full_raw.copy()
weather_full['datetime'] = pd.to_datetime(weather_full.apply(lambda row: f'{row['date']} {row['time']}', axis=1), format='%Y/%m/%d %R')
weather_full = weather_full.drop(['date', 'time'], axis='columns')
weather_full['vpd'] = calculate_VPD_series(weather_full, 'mean_T', 'mean_RH')
weather_full
```

```python
grouped_weather = group_vpd_by_date(weather_full)

# Drop outliers
grouped_weather = grouped_weather[~(grouped_weather['avg_vpd'] > 50)]
grouped_weather
```

```python
median_avg_vpd_full = calculate_median_df(grouped_weather, "avg_vpd", list(range(1997, 2026)))
median_avg_vpd_full
```

```python
fig_2010 = create_annual_lineplot(grouped_weather, vpd_column='avg_vpd', year_list=list(range(2010,2025 + 1)), highlight_year='2021')
fig_2010.update_layout(title='Average Daily VPD, 2010-2025', xaxis_title='Day of Year', yaxis_title='Average VPD')
fig_2010.show()
```

```python
fig_2021 = create_annual_lineplot(grouped_weather, vpd_column='avg_vpd', year_list=[2021], highlight_year='2021')
fig_2021.add_scatter(
    x=median_avg_vpd_full['DOY'],
    y=median_avg_vpd_full['doy_median'],
    mode='lines',
    line=dict(color="blue", width=2, dash="dash"),
    name='median'
)
fig_2021.update_layout(title='Average Daily VPD, 2021 and 1997-2025 median', xaxis_title='Day of Year', yaxis_title='Average VPD')
fig_2021.show()
```
